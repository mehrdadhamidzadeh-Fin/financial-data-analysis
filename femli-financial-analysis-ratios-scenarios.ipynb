```python
import pandas as pd

# واحد صورت‌های مالی: میلیون ریال
# مبنای مقایسه: ۱۴۰۳ تجدیدارائه‌شده در گزارش ۱۴۰۴
data = {
    "سال": [1403, 1404],
    "درآمد عملیاتی": [1_789_892_699, 3_218_894_510],
    "بهای تمام‌شده": [806_422_709, 1_354_424_181],
    "سود ناخالص": [983_469_990, 1_864_470_329],
    "سود خالص": [779_773_874, 1_460_550_037],
    "دارایی جاری": [1_258_762_744, 2_222_444_657],
    "کل دارایی‌ها": [2_879_151_852, 4_985_804_911],
    "بدهی جاری": [878_842_280, 1_841_806_377],
    "کل بدهی‌ها": [934_764_626, 1_914_606_935],
    "حقوق مالکانه": [1_944_387_226, 3_071_197_976],
    "موجودی کالا": [311_731_608, 694_556_486],
    "دریافتنی تجاری و سایر": [603_228_877, 901_580_580],
    "وجه نقد": [64_075_948, 161_608_044],
    "جریان نقد عملیاتی": [819_677_371, 1_620_265_320],
}

df = pd.DataFrame(data).set_index("سال")
df
```

```python
# بررسی اینکه سود ناخالص با درآمد منهای بهای تمام‌شده برابر باشد
df["کنترل سود ناخالص"] = (
    df["درآمد عملیاتی"] - df["بهای تمام‌شده"] - df["سود ناخالص"]
)

df[["درآمد عملیاتی", "بهای تمام‌شده", "سود ناخالص", "کنترل سود ناخالص"]]
```

```python
# رشد ۱۴۰۴ نسبت به ۱۴۰۳
growth = (
    df.loc[1404] / df.loc[1403] - 1
) * 100

# حاشیه‌های سود
margins = pd.DataFrame(index=df.index)
margins["حاشیه سود ناخالص (%)"] = (
    df["سود ناخالص"] / df["درآمد عملیاتی"] * 100
)
margins["حاشیه سود خالص (%)"] = (
    df["سود خالص"] / df["درآمد عملیاتی"] * 100
)

print("رشد ۱۴۰۴ نسبت به ۱۴۰۳:")
display(growth.to_frame("رشد (%)").round(2))

print("حاشیه‌های سود:")
display(margins.round(2))
```

```python
liquidity = pd.DataFrame(index=df.index)

# سرمایه در گردش = دارایی جاری - بدهی جاری
liquidity["سرمایه در گردش (میلیون ریال)"] = (
    df["دارایی جاری"] - df["بدهی جاری"]
)

# نسبت جاری = دارایی جاری / بدهی جاری
liquidity["نسبت جاری"] = (
    df["دارایی جاری"] / df["بدهی جاری"]
)

# نسبت آنی = (دارایی جاری - موجودی کالا) / بدهی جاری
liquidity["نسبت آنی"] = (
    (df["دارایی جاری"] - df["موجودی کالا"]) / df["بدهی جاری"]
)

# نسبت نقدی = وجه نقد / بدهی جاری
liquidity["نسبت نقدی"] = (
    df["وجه نقد"] / df["بدهی جاری"]
)

display(liquidity.round(2))
```

```python
leverage = pd.DataFrame(index=df.index)

# نسبت بدهی = کل بدهی‌ها / کل دارایی‌ها
leverage["نسبت بدهی (%)"] = (
    df["کل بدهی‌ها"] / df["کل دارایی‌ها"] * 100
)

# نسبت حقوق مالکانه به دارایی‌ها
leverage["حقوق مالکانه به دارایی‌ها (%)"] = (
    df["حقوق مالکانه"] / df["کل دارایی‌ها"] * 100
)

# نسبت بدهی به حقوق مالکانه
leverage["بدهی به حقوق مالکانه"] = (
    df["کل بدهی‌ها"] / df["حقوق مالکانه"]
)

display(leverage.round(2))
```

```python
# مانده‌های میانگین ۱۴۰۴
avg_assets_1404 = (
    df.loc[1403, "کل دارایی‌ها"] + df.loc[1404, "کل دارایی‌ها"]
) / 2

avg_inventory_1404 = (
    df.loc[1403, "موجودی کالا"] + df.loc[1404, "موجودی کالا"]
) / 2

avg_receivables_1404 = (
    df.loc[1403, "دریافتنی تجاری و سایر"]
    + df.loc[1404, "دریافتنی تجاری و سایر"]
) / 2

# گردش دارایی‌ها
asset_turnover_1404 = (
    df.loc[1404, "درآمد عملیاتی"] / avg_assets_1404
)

# گردش موجودی کالا و روزهای نگهداری موجودی (DIO)
inventory_turnover_1404 = (
    df.loc[1404, "بهای تمام‌شده"] / avg_inventory_1404
)
dio_1404 = 365 / inventory_turnover_1404

# گردش دریافتنی‌ها و روزهای وصول مطالبات (DSO)
receivables_turnover_1404 = (
    df.loc[1404, "درآمد عملیاتی"] / avg_receivables_1404
)
dso_1404 = 365 / receivables_turnover_1404

efficiency_1404 = pd.Series({
    "میانگین کل دارایی‌ها (میلیون ریال)": avg_assets_1404,
    "گردش دارایی": asset_turnover_1404,
    "گردش موجودی کالا": inventory_turnover_1404,
    "DIO (روز)": dio_1404,
    "گردش دریافتنی‌ها": receivables_turnover_1404,
    "DSO (روز)": dso_1404,
})

display(efficiency_1404.to_frame("۱۴۰۴").round(2))
```

```python
cash_quality = pd.DataFrame(index=df.index)

cash_quality["جریان نقد عملیاتی به سود خالص"] = (
    df["جریان نقد عملیاتی"] / df["سود خالص"]
)

display(cash_quality.round(2))
```

```python
# مقادیر را با ارقام گزارش، به میلیون ریال، جایگزین کن
capex = pd.Series({
    1403: 0,
    1404: 0,
})

cash_quality["مخارج سرمایه‌ای (میلیون ریال)"] = capex
cash_quality["جریان نقد آزاد (میلیون ریال)"] = (
    df["جریان نقد عملیاتی"] - capex
)
cash_quality["پوشش مخارج سرمایه‌ای"] = (
    df["جریان نقد عملیاتی"] / capex
)

display(cash_quality.round(2))
```

```python
# ارقام تلفیقی گزارش ۱۴۰۴؛ واحد: میلیون ریال
df.loc[1403, "جریان نقد عملیاتی"] = 785_588_772
df.loc[1404, "جریان نقد عملیاتی"] = 1_601_404_689

# CapEx = خرید دارایی‌های ثابت مشهود + نامشهود
capex = pd.Series({
    1403: 546_370_024 + 931_506,
    1404: 1_124_972_442 + 142_043,
})

cash_quality = pd.DataFrame(index=df.index)
cash_quality["جریان نقد عملیاتی / سود خالص"] = (
    df["جریان نقد عملیاتی"] / df["سود خالص"]
)
cash_quality["مخارج سرمایه‌ای (میلیون ریال)"] = capex
cash_quality["جریان نقد آزاد (میلیون ریال)"] = (
    df["جریان نقد عملیاتی"] - capex
)
cash_quality["پوشش مخارج سرمایه‌ای"] = (
    df["جریان نقد عملیاتی"] / capex
)

display(cash_quality.round(2))
```

```python
# محاسبه ROA و ROE برای سال ۱۴۰۴ بر پایه میانگین دارایی‌ها و حقوق مالکانه
avg_assets_1404 = (df.loc[1403, "کل دارایی‌ها"] + df.loc[1404, "کل دارایی‌ها"]) / 2
avg_equity_1404 = (df.loc[1403, "حقوق مالکانه"] + df.loc[1404, "حقوق مالکانه"]) / 2

profitability = pd.DataFrame(index=[1404])
profitability["میانگین کل دارایی‌ها"] = avg_assets_1404
profitability["میانگین حقوق مالکانه"] = avg_equity_1404
profitability["ROA (بازده دارایی‌ها)"] = (df.loc[1404, "سود خالص"] / avg_assets_1404) * 100
profitability["ROE (بازده حقوق مالکانه)"] = (df.loc[1404, "سود خالص"] / avg_equity_1404) * 100

display(profitability.round(2))
```

```python
# ارقام سود عملیاتی و هزینه مالی از صورت سود و زیان تلفیقی (میلیون ریال)
# سال ۱۴۰۳ تجدیدارائه‌شده و ۱۴۰۴ بر اساس گزارش تلفیقی ۱۴۰۴
ebit = pd.Series({
    1403: 948_665_448,
    1404: 1_826_976_769
})

interest_expense = pd.Series({
    1403: 4_797_528,
    1404: 10_397_405
})

debt_service = pd.DataFrame(index=df.index)
debt_service["سود عملیاتی (EBIT)"] = ebit
debt_service["هزینه‌های مالی"] = interest_expense
debt_service["پوشش بهره (مرتبه)"] = ebit / interest_expense
debt_service["نسبت هزینه مالی به درآمد (٪)"] = (interest_expense / df["درآمد عملیاتی"]) * 100

display(debt_service.round(2))
```

```python
# تجمیع شاخص‌های کلیدی عملکرد (KPI) در یک جدول جامع
kpi_summary = pd.DataFrame(index=[1403, 1404])

# ۱. حاشیه‌های سودآوری
kpi_summary["حاشیه ناخالص (٪)"] = (df["سود ناخالص"] / df["درآمد عملیاتی"]) * 100
kpi_summary["حاشیه خالص (٪)"] = (df["سود خالص"] / df["درآمد عملیاتی"]) * 100

# ۲. نقدینگی
kpi_summary["نسبت جاری"] = df["دارایی جاری"] / df["بدهی جاری"]
kpi_summary["نسبت آنی"] = (df["دارایی جاری"] - df["موجودی"]) / df["بدهی جاری"]

# ۳. اهرم مالی
kpi_summary["بدهی به دارایی (٪)"] = (df["کل بدهی‌ها"] / df["کل دارایی‌ها"]) * 100
kpi_summary["بدهی به حقوق مالکانه"] = df["کل بدهی‌ها"] / df["حقوق مالکانه"]

# ۴. پوشش و کیفیت جریان نقد
kpi_summary["CFO / سود خالص"] = cash_quality["جریان نقد عملیاتی / سود خالص"]
kpi_summary["پوشش CapEx"] = cash_quality["پوشش مخارج سرمایه‌ای"]
kpi_summary["پوشش بهره"] = debt_service["پوشش بهره (مرتبه)"]

# نمایش با دو رقم اعشار
display(kpi_summary.T.round(2))
```
